```python
x = torch.tensor(3.0, requires_grad= True)
y = x ** 2
```

```python
y
```

```python
y.backward()
```

```python
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
```

```python
y = w * x + b
z = y ** 2
```

```python
print(z.grad_fn)
```

```python
print(z.grad_fn.next_functions)
```

```python
z.backward()
```

```python
print(w.grad)
```

```python
print(x.grad)
```

```python
print(b.grad)
```
