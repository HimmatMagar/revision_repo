```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
z = w * x + b
```

```python
y_pred = torch.sigmoid(z)
```

```python
y_pred
```

```python
loss = torch.nn.functional.binary_cross_entropy(y_pred, y)
```

```python
loss
```

```python
loss.backward()
```

```python
w.grad
```

```python
b.grad
```
