```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
plt.plot(x, y)
plt.xlabel("X value")
plt.ylabel("Y value")
plt.title("Relation between X and Y")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 25]

plt.plot(x, y)
plt.xlabel("Days")
plt.ylabel("Values")
plt.title("Value over time");
```

```python
plt.scatter(x, y);
```

```python
categories = ["A", "B", "C", "D"]
values = [20, 35, 15, 40]

plt.bar(categories, values)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Category Comparison")

plt.show()
```

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 6]

plt.figure(figsize=(3, 3))
plt.hist(data)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(3, 3))
ax.plot(x, y)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("My Plot")
plt.show()
```

```python
fig, ax = plt.subplots(2, 2)

ax[0, 0].plot(x, y)
ax[0, 1].scatter(x, y)
ax[1, 0].bar(categories, values)
ax[1, 1].hist(data)

plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "age": [18, 20, 22, 25, 30],
    "salary": [20000, 25000, 30000, 40000, 50000]
})

fig, ax = plt.subplots(2, 2, figsize=(5, 5))
sns.scatterplot(data=df, x="age", y="salary", ax=ax[0, 0])
ax[0, 0].set_title("scatter plot")

sns.histplot(data=df, x="salary", kde=True,  ax=ax[0, 1])
ax[0, 1].set_title("Distribution")

# sns.boxplot(data=df, x="salary", ax=[1, 0])
# ax[1, 0].set_title("Box plot")

plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv("sales_data.csv")
df.head
```

```python
sns.set_theme(style="whitegrid")
```

```python
fig, ax = plt.subplots(3, 2, figsize=(12, 10))
fig.suptitle("Seaborn Statistical Plots", fontsize=15, fontweight="bold")

sns.histplot(data=df, x="sales", hue="category", kde=True, ax=ax[0, 0])
ax[0, 0].set_title("Sales Distribution by Category (with KDE)")

sns.boxplot(data=df, x="region", y="profit", hue="region", ax=ax[0, 1], legend=True)
ax[0, 1].set_title("Profit Spread by Region (Boxplot)")

sns.histplot(data=df, x="profit", hue="region", legend=True, kde=True, ax=ax[1, 0])
ax[1, 0].set_title("Profile distribution")

corr = df[["units_sold", "sales", "profit"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0, ax=ax[1, 1], fmt=".2f")
ax[1, 1].set_title("Correlation Heatmap")

sns.barplot(data=df, x="region", y="sales", hue="category", ax=ax[2, 0], errorbar=None)
ax[2, 0].set_title("Avg Daily Sales: Region x Category")
ax[2, 0].tick_params(axis="x", rotation=15)
ax[2, 0].legend(fontsize=8)

sns.scatterplot(data=df, x="sales", y="profit", ax=ax[2, 1], markers='o')
ax[2, 1].set_title("Scatter plot between sales and profit")

plt.tight_layout()
plt.show()
```
