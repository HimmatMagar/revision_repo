```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
plt.plot(x, y)
plt.xlabel("X value")
plt.ylabel("Y value")
plt.title("Relation between X and Y")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 25]

plt.plot(x, y)
plt.xlabel("Days")
plt.ylabel("Values")
plt.title("Value over time");
```

```python
plt.scatter(x, y);
```

```python
categories = ["A", "B", "C", "D"]
values = [20, 35, 15, 40]

plt.bar(categories, values)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Category Comparison")

plt.show()
```

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 6]

plt.hist(data)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(3, 3))
ax.plot(x, y)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("My Plot")
plt.show()
```

```python
fig, ax = plt.subplots(2, 2)

ax[0, 0].plot(x, y)
ax[0, 1].scatter(x, y)
ax[1, 0].bar(categories, values)
ax[1, 1].hist(data)

plt.show()
```
