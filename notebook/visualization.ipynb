```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
plt.plot(x, y)
plt.xlabel("X value")
plt.ylabel("Y value")
plt.title("Relation between X and Y")
plt.show()
```
